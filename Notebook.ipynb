```python
%%manim -ql CircleAndSqaure
class CircleAndSqaure(Scene):
    def construct(self):
        circle = Circle(radius=1)
        sq = Square()
        sq.rotate(PI/4)
        sq.next_to(circle,DOWN,buff=0)
        self.play(Create(sq))
        self.play(Create(circle))
        self.play(circle.animate.set_fill(GREEN,opacity=0.5))
```

```python
%%manim -ql CircleAndSqaureRotating
class CircleAndSqaureRotating(Scene):
    def construct(self):
        grp = VGroup()
        circle = Circle(radius=1)
        sq = Square()
        sq.rotate(PI/4)
        sq.next_to(circle,DOWN,buff=0)
        grp.add(*[circle,sq])
        self.play(Create(grp),run_time = 2)
        self.play(circle.animate.set_fill(GREEN,opacity=0.5))
        self.play(Rotate(grp,angle=2*PI,about_point=ORIGIN,axis=OUT,run_time = 5))
```

```python
%%manim -ql CircleAndSqaureRotatingWithTracing 
class CircleAndSqaureRotatingWithTracing(Scene):
    def construct(self):
        grp = VGroup()
        circle = Circle(radius=1)
        sq = Square()
        sq.rotate(PI/4)
        sq.next_to(circle,DOWN,buff=0)
        grp.add(*[circle,sq])

        path = TracedPath(sq.get_center,stroke_color=BLUE,stroke_width=2)
        self.add(path)

        self.play(Create(grp),run_time = 2)
        self.play(circle.animate.set_fill(GREEN,opacity=0.5))
        self.play(Rotate(grp,angle=2*PI,about_point=ORIGIN,axis=OUT,run_time = 5))

        self.play(FadeOut(sq))
```

```python
%%manim -ql CircleAndCircleRotatingAndMoved 
class CircleAndCircleRotatingAndMoved(Scene):
    def construct(self):
        grp = VGroup()
        c1 = Circle(radius=0.5)
        c1.set_fill(GREEN,opacity = 0.6)
        c2  = Circle(radius=0.7)
        c2.set_fill(YELLOW,opacity = 0.6)
        c1.next_to(c2,LEFT)

        grp.add(*[c1,c2])
        grp.shift(5*LEFT)
    
        self.play(Create(grp),run_time = 2)
        self.play(Rotate(grp,angle=2*PI,about_point=c2.get_center(),axis=OUT,run_time = 5))

        self.play(FadeOut(c2))
        self.play(FadeOut(c1))
```

```python
%%manim -ql CircleAndDotRotatingMovingANDTracing 
class CircleAndDotRotatingMovingANDTracing(Scene):
    def construct(self):
        grp = VGroup()
        c = Circle(radius=0.5).shift(LEFT*5)
        dot = Dot(c.get_left())
        grp.add(*[c,dot])
        trace = TracedPath(dot.get_center)
        
        grp.add_updater(lambda obj,dt: obj.rotate(3*dt))

        self.add(trace)
        self.play(FadeIn(grp))

        self.play(grp.animate.shift(RIGHT*10),run_time =7)
```

```python
%%manim -ql Snake
class Snake(Scene):
    def construct(self):
        tracker = ValueTracker(-2*PI)
        dot = always_redraw(lambda: Dot([
            tracker.get_value(),
            np.sin(tracker.get_value()),
            0
        ],color=YELLOW))
        tracer = TracedPath(dot.get_center,stroke_color=RED, stroke_width=9,dissipating_time=1.5)
        self.add(dot)
        self.add(tracer)
        self.play(tracker.animate.set_value(2*PI),run_time = 15)
```

```python
%%manim -ql EpiCycloids 
class EpiCycloids(Scene):
    def construct(self):
        r1 = 0.9
        r2 = 0.6
        f1 = 3
        f2 = 8

        c1 = Circle(radius=r1).shift([0,0,0])
        c2 = Circle(radius=r2).next_to(c1,LEFT,buff=0)
        dot = Dot(c2.get_left())
        trace = TracedPath(dot.get_center,dissipating_time=4)
        

        self.play(FadeIn(dot),run_time = 0.25)
        self.play(FadeIn(c2),run_time = 0.25)
        self.play(FadeIn(c1),run_time = 0.25)
        grp = VGroup(c2,dot)
        grp.add_updater(lambda obj,dt: obj.rotate(3*dt,about_point=c1.get_center()))
        dot.add_updater(lambda obj,dt: obj.rotate(8*dt,about_point=c2.get_center()))
        self.add(grp)
        self.add(trace)

        # self.play(grp.animate.shift(RIGHT*10),run_time =7)
        self.wait(15)
        grp.clear_updaters()
        dot.clear_updaters()
        self.play(FadeOut(grp), FadeOut(trace), run_time=0.5)
        self.play(FadeOut(dot),run_time = 0.5)
```

```python
%%manim -ql LissajousFigures
class LissajousFigures(Scene):
    def construct(self):
        w1 = 3
        w2 = 4
        phi = PI/3

        tracker = ValueTracker(-2*PI)
        dot = always_redraw(lambda: Dot([
            np.sin(w1*tracker.get_value()),
            np.sin(w2*tracker.get_value()+phi),
            0
        ],color=YELLOW))
        tracer = TracedPath(dot.get_center,stroke_color=RED, stroke_width=9,dissipating_time=None)
        self.add(dot)
        self.add(tracer)
        self.play(tracker.animate.set_value(2*PI),run_time = 20)
```

```python
%%manim -ql MyFlowerWIthColors
class MyFlowerWIthColors(Scene):
    def construct(self):
        colors= [PINK,RED,ORANGE,GOLD,YELLOW]
        k = 5
        scale = 3

        time_period = PI/k if k%2==1 else 2*PI/k
        tracker = ValueTracker(0)

        for x in range(k):
            _c = colors[x%len(colors)]
            dot = always_redraw(lambda: Dot([
                scale*np.sin(k*tracker.get_value())*np.cos(tracker.get_value()),
                scale*np.sin(k*tracker.get_value())*np.sin(tracker.get_value()),
                0
            ]))
            path = TracedPath(dot.get_center,stroke_width=9,stroke_color=_c)
            self.add(dot,path)
            self.play(tracker.animate.set_value((x+1)*time_period))
```

```python
%%manim -ql MyRotatingFlower
class MyRotatingFlower(Scene):
    def construct(self):
        colors= [PINK,RED,ORANGE,GOLD,YELLOW]
        k = 12
        scale = 3

        time_period = PI/k if k%2==1 else 2*PI/k
        tracker = ValueTracker(0)

        segments = VGroup()
        for x in range(k):
            _c = colors[x%len(colors)]
            dot = always_redraw(lambda: Dot([
                scale*np.sin(k*tracker.get_value())*np.cos(tracker.get_value()),
                scale*np.sin(k*tracker.get_value())*np.sin(tracker.get_value()),
                0
            ]))
            path = TracedPath(dot.get_center,stroke_width=9,stroke_color=_c)
            self.add(dot,path)
            segments.add(dot,path)
            self.play(tracker.animate.set_value((x+1)*time_period))
        flower = VGroup(*[VMobject(stroke_color=segment.stroke_color, stroke_width = segment.stroke_width).set_points(segment.get_points()) for segment in segments])
        self.remove(tracker, *segments)
        self.add(flower)

        self.play(Rotate(flower, angle = PI/2, about_point=ORIGIN,run_time=4))
```

```python
import cv2
import numpy as np
import json

def get_fourier_coefficients(image_path, num_samples=300, num_coeffs=60):
    # 1. Load image and extract edges/contours
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    _, thresh = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY_INV)
    contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)
    
    # Pick the longest continuous contour
    contour = max(contours, key=len).squeeze()  # Shape: (M, 2)
    
    # 2. Resample contour to evenly spaced points
    x, y = contour[:, 0], -contour[:, 1]  # Invert Y for Cartesian coordinates
    # Center and normalize coordinates
    x = (x - np.mean(x))
    y = (y - np.mean(y))
    scale = max(np.max(np.abs(x)), np.max(np.abs(y))) / 3.0  # Fit within Manim view
    x, y = x / scale, y / scale
    
    # Interpolate along path length
    distances = np.sqrt(np.diff(x, prepend=x[0])**2 + np.diff(y, prepend=y[0])**2)
    cumulative_dist = np.cumsum(distances)
    total_len = cumulative_dist[-1]
    
    interp_dists = np.linspace(0, total_len, num_samples, endpoint=False)
    x_sampled = np.interp(interp_dists, cumulative_dist, x)
    y_sampled = np.interp(interp_dists, cumulative_dist, y)
    
    z = x_sampled + 1j * y_sampled
    N = len(z)

    # 3. Compute Discrete Fourier Transform (DFT)
    coeffs = []
    freq_range = range(-num_coeffs // 2, num_coeffs // 2 + 1)
    
    for n in freq_range:
        # DFT: c_n = (1/N) * sum(z_k * exp(-i * 2*pi * n * k / N))
        c_n = np.sum(z * np.exp(-1j * 2 * np.pi * n * np.arange(N) / N)) / N
        coeffs.append({
            "freq": int(n),
            "radius": float(np.abs(c_n)),
            "phase": float(np.angle(c_n))
        })
        
    # Sort from largest radius to smallest for clean epicycle stacking
    coeffs.sort(key=lambda item: item["radius"], reverse=True)
    return coeffs
```

```python
%%manim -ql MyFourierEpicyle
class MyFourierEpicyle(Scene):
    def construct(self):
        coeffs = get_fourier_coefficients("doraemon.jpg")
        tracker = ValueTracker(0)

        def get_outine():
            curr_center = np.array([0,0,0])
            grp = VGroup()
            for coeff in coeffs:
                t = tracker.get_value()
                r = coeff['radius']
                w = coeff['freq']
                phi = coeff['phase']

                angle = w*t+phi
                next_center = np.array([
                    r*np.cos(angle),
                    r*np.sin(angle),
                    0
                ])+curr_center

                c = Circle(radius=r, color=BLUE, stroke_opacity = 0.4, stroke_width = 0.4).move_to(curr_center)
                v = Line(curr_center,next_center,color=WHITE, stroke_opacity = 0.6, stroke_width = 0.4)
                grp.add(c,v)
                curr_center = next_center
            tip = Dot(color=YELLOW,point=curr_center)
            grp.add(tip)
            return grp 

        outline = always_redraw(get_outine)

        dot = Dot().add_updater(lambda l: l.move_to(outline[-1].get_center()))
        trace = TracedPath(dot.get_center,stroke_width=9,stroke_color=GOLD)
        self.add(outline,dot,trace)
        self.play(tracker.animate.set_value(2*PI),run_time = 20)

```
